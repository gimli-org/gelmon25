We now look at the misfits. This should be available as separate function. The forward responses are contained in `self.responses` and the data in `self.DATA`, so we have to divide the difference by the error.

```python
misfits = np.log(xh2d.responses) - np.log(xh2d.DATA)
misfits /= np.reshape(xh2d.data["err"], [-1, 1])
chi2s = np.mean(misfits**2, axis=0)
plt.plot(xh2d.times, chi2s);
```
